```python
# surviving patients will have encounter_ids 1:20
surviving_ids = list(range(1, 21))

# non-surviving patients will have encounter_ids 21:40
nonsurviving_ids = list(range(21, 41))

# all trajectories will be 141 hours long
time_horizon = 143

# features will be 'hr', 'bp', 'spo2'
features = ['hr', 'bp', 'spo2']
```

```python
# create data frame with columns: encounter_id, time, hr, bp, spo2, mortality

# for surviving patients, hr will be 70 and will gradually increase to 100 over time_horizon
# for non-surviving patients, hr will be 70 and will gradually decrease to 40 over time_horizon

# for surviving patients, bp will be 110 and will gradually decrease to 90 over time_horizon
# for non-surviving patients, bp will be 120 and will gradually increase to 140 over time_horizon

# for surviving patients, spo2 will be 95 and will gradually increase to 100 over time_horizon
# for non-surviving patients, spo2 will be 95 and will gradually decrease to 80 over time_horizon

data = []
for eid in surviving_ids:
    for t in range(time_horizon):
        hr = 70 + (30 * t / time_horizon)
        bp = 110 - (20 * t / time_horizon)
        spo2 = 95 + (5 * t / time_horizon)
        data.append([eid, t, hr, bp, spo2, 0])  # mortality = 0 for surviving

for eid in nonsurviving_ids:
    for t in range(time_horizon):
        hr = 70 - (30 * t / time_horizon)
        bp = 120 + (20 * t / time_horizon)
        spo2 = 95 - (15 * t / time_horizon)
        data.append([eid, t, hr, bp, spo2, 1])  # mortality = 1 for non-surviving

columns = ['encounter_id', 'time', 'hr', 'bp', 'spo2', 'mortality']
artificial_df = pd.DataFrame(data, columns=columns)
```
