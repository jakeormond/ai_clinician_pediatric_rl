```python
import sys
import shutil
from pathlib import Path
import pickle as pkl
import os
import ipywidgets as widgets
from IPython.display import display

# resolve repo root whether notebook is run from analysis/ or project root
repo_root = Path.cwd()
if repo_root.name == "analysis":
    repo_root = repo_root.parent

# fallback to explicit root if needed
if not (repo_root / "src").exists():
    repo_root = Path("/home/jake/Documents/AI_Clinician/ai_clinician_pediatric_rl")

if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

# optional sanity check
import src.config_manager
print("Using src from:", src.config_manager.__file__)
```

```python
# generator so each folder is processed one manual cell-execution at a time
def folder_processor(model_folders):
    for model_folder in model_folders:
        print(f"Model folder: {model_folder}")

        cfg_path = os.path.join(temp_cfg_dir, model_folder, "config_manager.pkl")

        # load the config manager
        with open(cfg_path, "rb") as f:
            cfg_manager = pkl.load(f)

        print(f"Config manager loaded: {cfg_manager}")

        unit = cfg_manager.unit
        cohort = cfg_manager.cohort
        reward_cohort = cfg_manager.reward_cohort
        model_type = cfg_manager.model_type
        hospital = cfg_manager.hospital
        split_num = cfg_manager.split_num

        print(f"Model type: {model_type}, Unit: {unit}, Cohort: {cohort}, Reward Cohort: {reward_cohort.upper()}, Split Number: {split_num}")

        if reward_cohort != 'all' and reward_cohort != None:
            model_folder_new = f"{unit.lower()}_patients_health_score_from_{reward_cohort}_dataset"
        else:
            model_folder_new = f"{unit.lower()}_patients"

        if model_type == 'dynamic_deephit':
            model_folder_path = os.path.join('/home/jake/Documents/AI_Clinician/models/dynamic_deephit_results',
                                            hospital, 'five_split', f'split_{split_num}', 'sepsis', 'intake-vis',
                                            model_folder_new)
        else:
            model_folder_path = os.path.join("/home/jake/Documents/AI_Clinician/models/rl_models",
                                            hospital, model_type, f'split_{split_num}', 'sepsis', 'intake-vis',
                                            model_folder_new)

        print(f"Model folder path: {model_folder_path}")

        # check whether the model folder exists, otherwise create it
        if not os.path.exists(model_folder_path):
            os.makedirs(model_folder_path)
            print(f"Model folder created: {model_folder_path}")
        else:
            print(f"Model folder already exists: {model_folder_path}")

        # check if model_folder is already in model_folder_path, if so, skip moving it
        if os.path.exists(os.path.join(model_folder_path, model_folder)):
            print(f"Model folder {model_folder} already exists in {model_folder_path}, skipping move.")
        else:
            og_model_folder_path = os.path.join(temp_cfg_dir, model_folder)
            print(f"Moving model folder from {og_model_folder_path} to {model_folder_path}")
            shutil.move(og_model_folder_path, model_folder_path)

        yield model_folder
```

```python
temp_cfg_dir = "/home/jake/Documents/AI_Clinician/cfg_folder"

# find the folder nested inside, there is only one folder in the cfg_folder
model_folders = os.listdir(temp_cfg_dir)

_processor = folder_processor(model_folders)
```
