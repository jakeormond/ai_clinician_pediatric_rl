```python
import numpy as np
import torch
import pandas as pd
```

```python
gate_logit = torch.tensor([0.0, 1.0, -1.0])  # Example logits for gates
u = torch.rand_like(gate_logit).clamp(1e-6, 1 - 1e-6)
u
```

```python
gumbel_noise = torch.log(u) - torch.log(1 - u)
gumbel_noise
```

```python
gate_logit + gumbel_noise
```

```python
soft = torch.sigmoid((gate_logit + gumbel_noise) / 10)
soft
```

```python
soft = torch.sigmoid((gate_logit + gumbel_noise) / .3)
soft
```

```python
soft = torch.sigmoid((gate_logit + gumbel_noise) / .001)
soft
```

```python
soft = torch.sigmoid((gate_logit + gumbel_noise) / 1)
soft
```
