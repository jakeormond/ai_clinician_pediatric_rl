# load the pkl files, including config_manager to get run parameters

```python
import sys
# add project root (the directory that contains 'src')
sys.path.insert(0, "/home/jake/Documents/AI_Clinician/ai_clinician_pediatric_rl")


import pickle
from pathlib import Path
```

```python
experiment_dir = "/home/jake/Documents/AI_Clinician/experiments/"

run_name = "choc_basic_q_continuous_only_random_search_standard_20251014_003036"

fold_name = "fold_0_20251014_003036"

out_path = Path(f"{experiment_dir}/{run_name}/{fold_name}/run_params.txt")
```

```python
# load config_manager.pkl
config_manager_path = f"{experiment_dir}/{run_name}/{fold_name}/config_manager.pkl"

with open(config_manager_path, "rb") as f:
    config_manager = pickle.load(f)

config_manager.data_config
```

```python
vars_to_vars_to_write = {}

vars_to_vars_to_write['use_intermediate_rewards'] = config_manager.data_config['use_reward']
if vars_to_vars_to_write['use_intermediate_rewards']:
    vars_to_vars_to_write['reward_col'] = config_manager.data_config['reward_col']
    vars_to_vars_to_write['intermediate_reward_scale'] = config_manager.data_config['scale_reward']

vars_to_vars_to_write['use_terminal_reward'] = config_manager.data_config['use_terminal_reward']
if vars_to_vars_to_write['use_terminal_reward']:
    vars_to_vars_to_write['terminal_reward_col'] = config_manager.data_config['terminal_reward_col']
    vars_to_vars_to_write['scale_terminal_reward'] = config_manager.data_config['scale_terminal_reward']

vars_to_vars_to_write['min_traj_len'] = config_manager.data_config['min_traj_len']
vars_to_vars_to_write['max_traj_len'] = config_manager.data_config['max_trajectory_length']
```

```python
config_manager.data_config
```

```python
config_manager.training_config
```

```python
vars_to_vars_to_write['discount'] = config_manager.training_config['discount']
```

```python
with open(out_path, "w") as f:
    for var, val in vars_to_vars_to_write.items():
        f.write(f"{var}: {val}\n")
print(f"Wrote run parameters to {out_path}")
```
